## What is Pydantic?

Pydantic is the most widely used data validation and settings management library for python, utilizing type hints and ensure data structure integrety. It Validates, Parse and coerces data at runtime to match specified types, making it essential for building robust APIs and handling external data, with its core logic written in `Rust` for high performance.

## key Features and Benefits

1. __Data Validation and Parsing:__
    - Defines how data should be structured using standard python types, automatically enforcing these rules.
2. __Type Hint Integration:__
    - Uses Python's type annotations to define schemas, reducing the need for verbose validation code.
3. __Fast Performance:__
    - The code validation engine written in `Rust`, Making it extremely fast.
4. __Strict and Lax Modes:__
    - Supports both strict mode (enforcing strict types) and lax mode (attempting to coerce data, eg: converting `"1"` to `1`)
5. __Clear Error Handling:__
    - nProvides detailed errors when data validation fails.
6. __JSON Schema Generation:__
    - Pydantic models can easily generate JSON Schema for documentation.

## How to Use Pydantic

1. __Define Pydantic model__ that represents the __ideal schema__ for the data.
- This includes the expected fields, their types, and any validation constraints(eg: `gt=0` for positive numbers)

2. __Instantiate the model with raw input data__ (Usually a dictionary or JSON-like structure).
- Pydantic will automatically __Validates__ the data and __coerce__ it into correct Python types(if possible).
- If the data doesn't meet the model's requirements, Pydantic raises a `ValidationError`.

3. __Pass the validated model object__ to functions or use it throughout your codebase.
- This ensures that every part of your program works with __Clean, Type-safe, and logically valid data__. 

```python
from pydantic import BaseModel, ValidationError, Field
from typing import Dict, List, Optional

class PatientData(BaseModel):
    name: str = Field(..., description="The name of the patient.")
    age: int = Field(..., description="The age of the patient.",  gt=0, le=150)
    weight: float = Field(..., description="The weight of the patient in kilograms.", gt=0)
    married: bool = Field(..., description="Marital status of the patient.")
    allergies: Optional[List[str]] = Field(None, description="List of allergies the patient has.")
    contact_info: Dict[str, str] = Field(..., description="Contact information of the patient.")

def add_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print("Data `Added` Successfully to the database.")

def update_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print("Data `Updated` Successfully in the database.")

patient_info = {
    "name": "Chinta",
    "age": 25,
    "weight": 70.5,
    "married": True,
    "allergies": ["Peanuts", "Shellfish"],
    "contact_info": {
        "email": "chinta@example.com",
        "phone": "123-456-7890"
    }
}

patient = PatientData(**patient_info)
add_patient_data(patient)

patient_updated_info = {
    "name": "Chinta",
    "age": 26,
    "weight": 71.0,
    "married": True,
    "allergies": ["Peanuts", "Shellfish"],
    "contact_info": {
            "email": "chintaupdated@example.com",
            "phone": "123-456-7890"
        }
}
patient_updated = PatientData(**patient_updated_info)
update_patient_data(patient_updated)
```

# Computed Field

```python
from pydantic import BaseModel, ValidationError, Field, EmailStr, AnyUrl, field_validator, model_validator, computed_field
from typing import Dict, List, Optional, Annotated

class PatientData(BaseModel):
    name: Annotated[str, Field(..., description="The name of the patient.", min_length=1, max_length=100, example="John Doe")]
    linkedin_url: Annotated[AnyUrl, Field(..., description="LinkedIn URL of the patient.")]
    age: Annotated[int, Field(..., description="The age of the patient.",  gt=0, le=150)]
    weight: Annotated[float, Field(..., strict=True, description="The weight of the patient in kilograms.", gt=0)]
    height: Annotated[Optional[float], Field(None, description="The height of the patient in centimeters.", gt=0)]
    married: Annotated[bool, Field(..., description="Marital status of the patient.")]
    allergies: Annotated[Optional[List[str]], Field(None, description="List of allergies the patient has.")]
    contact_info: Annotated[Dict[str, str], Field(..., description="Contact information of the patient.")]
    email: Annotated[Optional[EmailStr], Field(None, description="Email address of the patient.")]

    @field_validator('age')
    def validate_age(cls, value):
        if value < 0:
            raise ValueError("Age cannot be negative.")
        return value

    @field_validator('email')
    @classmethod
    def validate_email(cls, value):
        if value and "@" not in value:
            raise ValueError("Invalid email address.")
        valid_email_domains = ['hdfc.com', 'tcs.com'] 
        domain_name = value.split('@')[-1] if value else None
        if domain_name and domain_name not in valid_email_domains:
            raise ValueError(f"Email domain must be one of the following: {', '.join(valid_email_domains)}")
        return value

    @field_validator('name')
    @classmethod
    def validate_name(cls, value):
        if not value.strip():
            raise ValueError("Name cannot be empty or whitespace.")
        return value.upper()

    @model_validator(mode='after')
    @classmethod
    def validate_emergency_model(cls, model):
        if model.age > 60 and 'emergency' not in model.contact_info:
            raise ValueError("Patients over 60 must have an emergency contact in their contact_info.")
        return model

    @computed_field
    @property
    def bmi(self) -> Optional[float]:
        if self.height and self.weight:
            height_in_meters = self.height / 100
            return self.weight / (height_in_meters ** 2)
        return None

def add_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.linkedin_url)
    print(patient.age)
    print(patient.weight)
    print(patient.height)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print(f"BMI {patient.bmi:.2f}")
    print("Data `Added` Successfully to the database.")

def update_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.linkedin_url)
    print(patient.age)
    print(patient.weight)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print(f"BMI {patient.bmi:.2f}")
    print("Data `Updated` Successfully in the database.")

patient_info = {
    "name": "Chinta",
    "linkedin_url": "https://www.linkedin.com/in/chintamani-pala",
    "age": 80,
    "weight": 70.5,
    "height": 175.0,
    "married": True,
    "allergies": ["Peanuts", "Shellfish"],
    "contact_info": {
        "phone": "123-456-7890",
        "emergency": "987-654-3210"
    },
    "email": "chinta@hdfc.com"
}

patient = PatientData(**patient_info)
add_patient_data(patient)

patient_updated_info = {
    "name": "Chinta",
    "linkedin_url": "https://www.linkedin.com/in/chintamani-pala",
    "age": 26,
    "weight": 71.0,
    "height": 175.0,
    "married": True,
    "allergies": ["Peanuts", "Shellfish"],
    "contact_info": {
            "phone": "123-456-7890",
            "emergency": "987-654-3210"
    },
    "email": "chintaupdated@tcs.com",
}
patient_updated = PatientData(**patient_updated_info)
update_patient_data(patient_updated)
```
